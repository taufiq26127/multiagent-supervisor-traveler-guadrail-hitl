````python
data_6 = {'error': True, 'message': 'API rate limit exceeded', 'code': 429}

# 1. Loop KEY saja (default)
for h in data_6:
    print(h)                    # -> "error", "message", "code"

# 2. Loop KEY saja (eksplisit, sama persis hasilnya seperti di atas)
for h in data_6.keys():
    print(h)                    # -> "error", "message", "code"

# 3. Loop VALUE saja
for h in data_6.values():
    print(h)                    # -> True, "API rate limit exceeded", 429

# 4. Loop KEY dan VALUE bersamaan
for k, v in data_6.items():
    print(f"{k}: {v}")          # -> "error: True", "message: API rate limit exceeded", "code: 429"
````

````python
def ambil_results_aman(data):
    if data is None:
        return []
    if isinstance(data, list):
        return data
    if isinstance(data, dict):
        return data.get("results") or []   # "or []" -> jaga2 kalau None
    return []

for label, contoh in [("8a (None)", soal_8a), ("8b (list kosong)", soal_8b), ("8c (results=None)", soal_8c)]:
    hasil = ambil_results_aman(contoh)
    print(f"  {label}: aman, hasil = {hasil} (tidak crash, {len(hasil)} item)")
print("SOLUSI: fungsi ambil_results_aman() dengan .get(key) or [] mencegah crash di semua kasus.")
````

````python
soal_9 = [
    {
        "type": "text",
        # text ini isinya JSON string LAGI yang isinya text lagi (kasus tool-chaining)
        "text": json.dumps([
            {"type": "text", "text": '{"city": "Bali", "temp_c": 30}', "id": "inner_1"}
        ]),
        "id": "outer_1",
    }
]

soal_9
````

````python
def unwrap_berulang(raw, max_depth=5):
    """Bongkar wrapper MCP berkali-kali sampai ketemu data bersih."""
    current = raw
    for _ in range(max_depth):
        if isinstance(current, str):
            try:
                current = json.loads(current)
                continue
            except json.JSONDecodeError:
                return current  # string biasa, bukan JSON -> berhenti
        if isinstance(current, list) and len(current) > 0:
            first = current[0]
            if isinstance(first, dict) and "text" in first and set(first.keys()) <= {"type", "text", "id"}:
                current = first["text"]
                continue
        # sudah bukan wrapper lagi -> berhenti
        break
    return current
````

````python
unwrapped_9 = unwrap_berulang(soal_9)
print(unwrapped_9)
````

````python
"""
Contoh kasus sederhana: kenapa perlu 'ambil_aman' sebelum loop data
"""

def ambil_aman(data):
    if data is None:
        return []
    if not isinstance(data, list):
        return []
    return data


def garis(judul):
    print(f"\n{'='*55}\n{judul}\n{'='*55}")


# ============================================================
# KASUS 1: Server hotel lagi down, hasilnya None
# ============================================================
garis("KASUS 1: Server hotel down -> hasilnya None")

hasil_dari_server = None   # ini yang dikembalikan API pas server error

print("Kalau LANGSUNG di-loop tanpa cek:")
print("""
    for hotel in hasil_dari_server:
        print(hotel["name"])
""")
print(">> CRASH: TypeError: 'NoneType' object is not iterable")

print("\nKalau pakai ambil_aman() dulu:")
data_aman = ambil_aman(hasil_dari_server)
print(f"data_aman = {data_aman}")
for hotel in data_aman:
    print(hotel["name"])
print(">> AMAN, gak nge-print apa-apa (karena memang gak ada hotel), tapi TIDAK CRASH")


# ============================================================
# KASUS 2: Server hotel jawab, tapi search-nya gak nemu apa-apa
# ============================================================
garis("KASUS 2: Pencarian hotel tidak ketemu -> hasilnya []")

hasil_dari_server_2 = []   # server jawab normal, cuma memang kosong

print("Kalau LANGSUNG di-loop:")
for hotel in hasil_dari_server_2:
    print(hotel["name"])
print(">> AMAN-AMAN AJA sebenarnya, karena [] itu sudah list (bisa langsung di-loop)")

print("\nTapi tetap aman juga kalau dilewatin ambil_aman() (gak ngerusak apa-apa):")
data_aman_2 = ambil_aman(hasil_dari_server_2)
print(f"data_aman_2 = {data_aman_2}")


# ============================================================
# KASUS 3: Server malah balikin pesan ERROR (dict), bukan list hotel
# ============================================================
garis("KASUS 3: Server balikin dict error, bukan list hotel")

hasil_dari_server_3 = {"error": "Rate limit exceeded"}

print("Kalau LANGSUNG di-loop:")
print("""
    for hotel in hasil_dari_server_3:
        print(hotel["name"])
""")
print(">> CRASH beda: 'h' jadi STRING key ('error'), bukan dict hotel")
print("   -> TypeError: string indices must be integers (pas akses hotel['name'])")

print("\nKalau pakai ambil_aman() dulu:")
data_aman_3 = ambil_aman(hasil_dari_server_3)
print(f"data_aman_3 = {data_aman_3}")
for hotel in data_aman_3:
    print(hotel["name"])
print(">> AMAN, karena dict bukan list -> ambil_aman() ganti jadi [] duluan")


# ============================================================
# KASUS 4: Server jawab BENAR, ada hotelnya
# ============================================================
garis("KASUS 4: Data normal, ada isinya")

hasil_dari_server_4 = [
    {"name": "Hotel Mulia"},
    {"name": "Hotel Aston"},
]

print("Kalau LANGSUNG di-loop:")
for hotel in hasil_dari_server_4:
    print(hotel["name"])
print(">> AMAN, karena memang sudah list of dict yang benar")

print("\nKalau pakai ambil_aman() dulu (hasilnya SAMA, tidak ada yang rusak):")
data_aman_4 = ambil_aman(hasil_dari_server_4)
for hotel in data_aman_4:
    print(hotel["name"])
print(">> Tetap jalan normal, ambil_aman() gak ganggu data yang memang sudah benar")


garis("KESIMPULAN")
print("""
ambil_aman() itu kayak SATPAM di pintu masuk sebelum loop:
- Kalau None       -> ditolak, dikasih [] (kosong, aman)
- Kalau bukan list -> ditolak, dikasih [] (kosong, aman)
- Kalau list benar -> dibolehin lewat apa adanya

Efeknya: loop di bawahnya GAK PERNAH CRASH, paling parah cuma 
gak nge-print apa-apa (karena datanya memang kosong/rusak).
""")
````

````python
text = 'Tentu, ini hasilnya:\n```json\n{"name": "Hotel Mulia", "price": 1200000}\n```\nSemoga membantu!'

start = text.find("{")   # -> index tempat { pertama muncul
end = text.rfind("}")    # -> index tempat } terakhir muncul

print(text)

potongan = text[start : end + 1]
print("Potongan JSON:", potongan)
# -> '{"name": "Hotel Mulia", "price": 1200000}'
print(type(potongan))

hasil = json.loads(potongan)
print(hasil)
# -> {'name': 'Hotel Mulia', 'price': 1200000}
print(type(hasil))
# -> <class 'dict'>
````
